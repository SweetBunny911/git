```python
import pandas as pd
# Состояния
states = [
    "Создан",
    "Оплачен",
    "В обработке",
    "Собран",
    "Отправлен",
    "Доставлен",
    "Получен",
    "Отменен",
    "Возвращен"
]

# Матрица переходов (инициализация всех значений как 'Невозможно')
transition_matrix = pd.DataFrame("Невозможно", index=states, columns=states)

# Условия переходов
transition_matrix.at["Создан", "Оплачен"] = "Оплата получена"
transition_matrix.at["Оплачен", "В обработке"] = "Заказ обрабатывается"
transition_matrix.at["В обработке", "Собран"] = "Заказ собран"
transition_matrix.at["Собран", "Отправлен"] = "Заказ отправлен"
transition_matrix.at["Отправлен", "Доставлен"] = "Заказ доставлен"
transition_matrix.at["Доставлен", "Получен"] = "Получен клиентом"

# Переходы для отмены
transition_matrix.at["Создан", "Отменен"] = "Заказ отменен"
transition_matrix.at["Оплачен", "Отменен"] = "Заказ отменен"
transition_matrix.at["В обработке", "Отменен"] = "Заказ отменен"

# Переходы для возврата
transition_matrix.at["Отправлен", "Возвращен"] = "Возврат инициирован"
transition_matrix.at["Доставлен", "Возвращен"] = "Возврат инициирован"
transition_matrix.at["Получен", "Возвращен"] = "Возврат инициирован"

# Переходы для "Возвращен"
transition_matrix.at["Возвращен", "Создан"] = "Возврат принят"

# Сохранение таблицы в Excel
file_path = "transition_matrix_order_states.xlsx"
transition_matrix.to_excel(file_path)

print(f"Матрица состояний успешно сохранена в файл: {file_path}")
```
